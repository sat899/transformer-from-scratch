Imports

```python
import numpy as np
```

Simple Example (4 Inputs, 3 Features, 5 Classes)

```python
# Example with 4 inputs and 3 features

input_1 = [2.0, 1.0, 0.1]
input_2 = [-1.2, 3.5, -0.5]
input_3 = [1.8, 1.5, -2.0]
input_4 = [0.5, -1.0, 2.0]

simple_input = np.array([input_1, input_2, input_3, input_4])  # shape (4, 3)
simple_input, simple_input.shape
```

```python
# Linear layer / weight matrix is 3 x 5 (mapping 3 input features to 5 output classes)
simple_weights = np.random.randn(3, 5)
simple_bias = np.zeros(5)
simple_weights, simple_bias
simple_weights.shape, simple_bias.shape

simple_input.shape, simple_weights.shape, simple_bias.shape
```

```python
# Output has 4 inputs and 5 classes, so the shape of the output is (4, 5)
# The labels must match the number of inputs, so we have 4 labels, one for each input.
# The labels can be represented in two ways: sparse (as integers) or one-hot encoded (as binary vectors).
simple_labels_sparse = np.array([0, 1, 2, 3])

simple_labels_one_hot = np.array([
    [1, 0, 0, 0, 0],  # True class is index 0
    [0, 1, 0, 0, 0],  # True class is index 1
    [0, 0, 1, 0, 0],   # True class is index 2
    [0, 0, 0, 1, 0]   # True class is index 3
])

simple_labels_sparse.shape, simple_labels_one_hot.shape
```
