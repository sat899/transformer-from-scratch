Cross-Entropy Loss

```python
from src.utils.optimisation import cross_entropy_loss_one_hot, cross_entropy_loss_sparse
```

```python
# Example with 3 samples and 3 classes

sample_1 = [0.70, 0.20, 0.10]
sample_2 = [0.10, 0.85, 0.05]
sample_3 = [0.60, 0.30, 0.10]

predictions = np.array([sample_1, sample_2, sample_3])
predictions, predictions.shape
```

```python
# labels
labels_sparse = np.array([0, 1, 2])

labels_one_hot = np.array([
    [1, 0, 0],  # True class is index 0
    [0, 1, 0],  # True class is index 1
    [0, 0, 1]   # True class is index 2
])

labels_sparse.shape, labels_one_hot.shape
```

```python
loss_one_hot = cross_entropy_loss_one_hot(predictions, labels_one_hot)
loss_one_hot
```

```python
loss_sparse = cross_entropy_loss_sparse(predictions, labels_sparse)
loss_sparse
```
